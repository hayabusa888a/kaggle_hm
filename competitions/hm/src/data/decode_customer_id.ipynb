提出用ファイルに変換するためのコード（customer_idをもとに戻す）

```python
# 変換対象のファイル名を入力（data/processed/ 以下のファイル名）
INPUT_FILENAME = "C:\kaggle_nvidia\kaggle_hm\competitions\hm\data\processed\sub_repeat_popularity.csv"
```

```python
import pandas as pd
from pathlib import Path

ROOT = Path("../..").resolve()
input_path = ROOT / "data" / "processed" / INPUT_FILENAME
map_path = ROOT / "data" / "processed" / "customer_id_map.csv"
output_path = ROOT / "outputs" / "submissions" / (Path(INPUT_FILENAME).stem + "_decoded.csv")

id_map = pd.read_csv(map_path).set_index("customer_id_int")["customer_id"]

df = pd.read_csv(input_path)
first_col = df.columns[0]
df[first_col] = df[first_col].map(id_map)

output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)
print(f"保存完了: {output_path}")
df.head(3)
```

```python
from pathlib import Path

# 提出ファイルのパスを指定

sub_path = "C:\kaggle_nvidia\kaggle_hm\competitions\hm\data\processed\sub_repeat_popularity.csv"

# 読み込み
submission_full = pd.read_csv(sub_path, dtype={"customer_id": "str"})

# article_idを0埋め10桁に変換
def pad_article_ids(prediction_str):
    return " ".join(str(aid).zfill(10) for aid in prediction_str.split())

submission_full["prediction"] = submission_full["prediction"].apply(pad_article_ids)

# 上書き保存
# submission_full.to_csv(sub_path, index=False)
save_path = 'C:\kaggle_nvidia\kaggle_hm\competitions\hm\data\processed\sub_repeat_popularity_2.csv'
submission_full.to_csv(save_path, index=False)

print("保存完了")
print(submission_full.head(3))
```
