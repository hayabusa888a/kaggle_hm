```python
# t_datをdatetime型に変換
transactions["t_dat"] = pd.to_datetime(transactions["t_dat"])
# 基準日（データの最終日）を確認
print(transactions["t_dat"].max())  # 2020-09-19が出るはず
print(transactions["t_dat"].min())  # 2018-09-20が出るはず
```

```python
# 基準日を設定（データの最終日）
reference_date = transactions["t_dat"].max()

# 各トランザクションが「最終日から何週前か」を週番号として割り振る
# 最新が週番号が大きくなるように設定
transactions["week"] = (
    (transactions["t_dat"] - transactions["t_dat"].min()).dt.days // 7
)

# week列を確認
print(transactions["week"].min())
print(transactions["week"].max())
print(transactions[["t_dat", "week"]].head(10))
```

```python
# valid用の設定
# x:week104より前の購買履歴
# y:week104以降の購買履歴
valid_x = transactions[transactions["week"] <104]
valid_y = transactions[transactions["week"] == 104]

print(f"valid_x: {len(valid_x)}行")
print(f"valid_y: {len(valid_y)}行")
print(f"valid_yのユニークユーザー数: {valid_y['customer_id'].nunique()}")
print(f"valid_yのユニーク商品数: {valid_y['article_id'].nunique()}")
```

```python
# 週番号を付与したtransactionsを保存
transactions.to_csv(
    "C:/kaggle_nvidia/kaggle_hm/competitions/hm/data/processed/transactions_with_week.csv",
    index=False
)

# valid_x・valid_yを保存
valid_x.to_csv(
    "C:/kaggle_nvidia/kaggle_hm/competitions/hm/data/processed/valid_x.csv",
    index=False
)

valid_y.to_csv(
    "C:/kaggle_nvidia/kaggle_hm/competitions/hm/data/processed/valid_y.csv",
    index=False
)

print("保存完了")
```
