```python
import pandas as pd
from pathlib import Path

ROOT = Path("../..").resolve()
PROCESSED_DIR = ROOT / "data" / "processed"
```

```python
# 購入回数上位12件の article_id を特定
transactions = pd.read_csv(
    PROCESSED_DIR / "transactions_converted.csv",
    dtype={"article_id": str}
)

top12 = (
    transactions
    .groupby("article_id")
    .size()
    .nlargest(12)
    .index
    .tolist()
)
prediction_str = " ".join(top12)
print(f"top12 article_ids: {prediction_str}")
```

```python
# 全 customer_id に同じ prediction を割り当てて baseline.csv を作成
customers = pd.read_csv(PROCESSED_DIR / "customers_converted.csv", usecols=["customer_id"])

baseline = pd.DataFrame({
    "customer_id": customers["customer_id"],
    "prediction": prediction_str
})

baseline.to_csv(PROCESSED_DIR / "baseline.csv", index=False)
print(f"保存完了: {PROCESSED_DIR / 'baseline.csv'}")
print(f"rows: {len(baseline):,}")
baseline.head(3)
```
